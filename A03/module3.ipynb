```python
# Loading the gate class and checking the weighted sum for inputs 1 and 1
import numpy as np
from logic_gate import LogicGate

gate = LogicGate()
x = np.array([1, 1])
w = np.array([0.5, 0.5])
th = 0.7
print(x * w)
print(np.sum(x * w))
print(np.sum(x * w) > th)
```

## and / nand / or / nor

and: 

i used w = 0.5, 0.5 and th = 0.7. (1,1) sums to 1.0 which passes 0.7. one 1 only gets to 0.5 so it stays 0. the slide also had 1, 1, th 1.9 and 0.5, 0.2, th 0.6. those work too, i just didnt use them.

nand: 

the picture has weights -2 and -2 and a 3 in the neuron
3 is the bias b. 
class formula when the weighted sum + b > 0, so th = -3. (1,1) is -4 
which is not bigger than -3.

or: 

slide example w = 0.5, 0.5, th = 0. any 1 gives 0.5. (0,0) is 0
0 is not greater than 0

nor: 

same perceptron, different weights, flipped the or weights to -0.5, -0.5 and set th = -0.5, so only (0,0) can pass

## xor

```python
# Print OR, NAND, and the AND of those two, which is the XOR result.
print("x1 x2   or  nand  xor")
for x1 in [0, 1]:
    for x2 in [0, 1]:
        or_y = gate.or_gate(x1, x2)
        nand_y = gate.nand_gate(x1, x2)
        print(x1, x2, "  ", or_y, "  ", nand_y, "   ", gate.and_gate(or_y, nand_y))
```

## tables

same functions module3.py prints

```python
# Print every input pair for and, nand, or, nor, and xor.
for name, func in [
    ("and", gate.and_gate),

    ("nand", gate.nand_gate),

    ("or", gate.or_gate),

    ("nor", gate.nor_gate),

    ("xor", gate.xor_gate),
]:
    print(name)
    for x1 in [0, 1]:
        for x2 in [0, 1]:
            print(x1, x2, func(x1, x2))
    print()
```
