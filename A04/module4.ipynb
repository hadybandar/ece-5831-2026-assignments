# module 4

multilayer perceptron with forward propagation, same idea as the class example (2 -> 3 -> 2 -> 2) but bigger.

network:
- input layer: 3 neurons
- layer 1: 5 neurons, sigmoid
- layer 2: 5 neurons, sigmoid
- layer 3: 2 neurons, sigmoid
- layer 4 (output): 2 neurons, relu

weight shapes go (neurons in previous layer) x (neurons in next layer), so w1 is 3x5, w2 is 5x5, w3 is 5x2, w4 is 2x2. each bias has one value per neuron in its layer. the weight values are just ones i picked, nothing is trained.

```python
# Load the class and print the shape of every weight and bias
import numpy as np
from multilayer_perceptron import MultilayerPerceptron
mlp = MultilayerPerceptron()
for key, value in mlp.network.items():
    print(key, value.shape)
```

## forward propagation step by step

each layer does a = z_prev @ w + b and then the activation function. layers 1-3 use sigmoid, layer 4 uses relu so the output can never be negative.

```python
# Push one input through each layer by hand and print what every layer outputs
x = np.array([1.0, 0.5, -1.0])
net = mlp.network
a1 = np.dot(x, net["w1"]) + net["b1"]
z1 = mlp.sigmoid(a1)
a2 = np.dot(z1, net["w2"]) + net["b2"]
z2 = mlp.sigmoid(a2)
a3 = np.dot(z2, net["w3"]) + net["b3"]
z3 = mlp.sigmoid(a3)
a4 = np.dot(z3, net["w4"]) + net["b4"]
y = mlp.relu(a4)
print("input:            ", x)
print("output of layer 1:", np.round(z1, 4))
print("output of layer 2:", np.round(z2, 4))
print("output of layer 3:", np.round(z3, 4))
print("layer 4 before relu:", np.round(a4, 4))
print("output (relu):    ", np.round(y, 4))
```

```python
y_class = mlp.forward(x)
print("mlp.forward(x):", np.round(y_class, 4))
print("same as manual:", np.allclose(y, y_class))
```

## more inputs

some inputs give 0 on the second output, that is relu cutting off the negative value.

```python
inputs = [
    [0.0, 0.0, 0.0],
    [1.0, 0.5, -1.0],
    [0.2, -0.7, 0.9],
    [-1.0, 2.0, 0.5],
    [3.0, -2.0, 1.0],
]
for x in inputs:
    print(x, "->", np.round(mlp.forward(np.array(x)), 4))
```
